## 7)Test the Analysis of Variance between etest_p and mba_p at signifance level 5%.(Make decision using Hypothesis Testing)

```python
# since the question is very straight we can directly performe "Hypothesis testing" 

# one way ANOVA
```

```python
import scipy.stats as stats

stats.f_oneway(dataset['etest_p'],dataset['mba_p'])
```

```python
# since the pvalue(4.6725) > 0.05 -----> We accept null hypothesis and reject alternative 

# Null Hypothesis(H0) ---> etest_p = mba_p
```

## 8)Test the similarity between the degree_t(Sci&Tech) and specialisation(Mkt&HR) with respect to salary at significance level of 5%.(Make decision using Hypothesis Testing)


```python
# since the question is very straight we can directly performe "Hypothesis testing" 

# two way ANOVA
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Example: Testing if Percentage Score depends on Gender AND Degree Type
model = ols('salary ~ C(degree_t) + C(specialisation) + C(degree_t):C(specialisation)', data=dataset).fit()

anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

```python
# If p-value < 0.05 → Reject H₀ → Significant effect on salary.
# If p-value ≥ 0.05 → Fail to reject H₀ → No significant effect.
```

```python
# Output:

# degree_t - p value = 0.201122 ----> p > 0.05 = H0 - accepted ---> No significant effect
# specialisation - p value = 0.000051 ----> p < 0.05 = H0 - rejected -----> Significant effect on salary
# C(degree_t):C(specialisation) ---> Interaction Effect: Does the effect of degree_t depend on the specialisation stream?
# p value = 0.781152 ----> p > 0.05 = H0 - accepted ------> No significant effect
```

# 10)What is the probability Density Function of the salary range from 700000 to 900000?

```python
def get_pdf_probability(dataset,startrange,endrange):
    from matplotlib import pyplot
    from scipy.stats import norm
    import seaborn as sns
    ax = sns.distplot(dataset,kde=True,kde_kws={'color':'blue'},color='Green')
    pyplot.axvline(startrange,color='Red')
    pyplot.axvline(endrange,color='Red')
    # generate a sample
    sample = dataset
    # calculate parameters
    sample_mean =sample.mean()
    sample_std = sample.std()
    print('Mean=%.3f, Standard Deviation=%.3f' % (sample_mean, sample_std))
    # define the distribution
    dist = norm(sample_mean, sample_std)
    
    # sample probabilities for a range of outcomes
    values = [value for value in range(startrange, endrange)]
    probabilities = [dist.pdf(value) for value in values]    
    prob=sum(probabilities)
    print("The area between range({},{}):{}".format(startrange,endrange,sum(probabilities)))
    return prob
```

```python
get_pdf_probability(dataset["salary"],700000,900000)
```

# 11)Test the similarity between the degree_t(Sci&Tech)with respect to etest_p and mba_p at significance level of 5%.(Make decision using Hypothesis Testing)

```python
# since the question is very straight we can directly performe "Hypothesis testing" 
```

```python
from scipy.stats import ttest_rel
etest_p = dataset[dataset['degree_t']=='Sci&Tech']['etest_p']
mba_p = dataset[dataset['degree_t']=='Sci&Tech']['mba_p']
ttest_rel(etest_p, mba_p)
```

```python
# since the pvalue(5.51792) > 0.05 -----> We accept null hypothesis and reject alternative 

# Null Hypothesis(H0) ---> There is no significant difference between etest_p and mba_p among Sci&Tech students.
```

# 12)Which parameter is highly correlated with salary?

```python
def QuanQual(dataset):
    Quan = []
    Qual = []
    for columnName in dataset.columns:
        if(dataset[columnName].dtype=='O'):
            Qual.append(columnName)
        else:
            Quan.append(columnName)
    return Quan,Qual
```
