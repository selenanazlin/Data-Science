```python
dataset = pd.read_csv("Preplacement.csv")
```

```python
dataset.drop('sl_no',inplace=True,axis=1)
```

```python
dataset
```

```python
def QuanQual(dataset):
    Quan = []
    Qual = []
    for columnName in dataset.columns:
        if(dataset[columnName].dtype=='O'):
            Qual.append(columnName)
        else:
            Quan.append(columnName)
    return Quan,Qual
```

```python
QuanQual(dataset)
```

```python
quan,qual = QuanQual(dataset)
```

```python
dataset[quan]
```

```python
dataset[qual]
```

```python
# diff group (Male,female) bt same condition (salary)
```

```python
from scipy.stats import ttest_rel
male = dataset[dataset['gender']=='M']['ssc_p']
male1 = dataset[dataset['gender']=='M']['hsc_p']
ttest_rel(male, male1)
```

```python
# Since the p value(0.494) is > 0.05 we accept null hypothesis and reject alternative hypothesis
```

```python
from scipy.stats import ttest_ind
male = dataset[dataset['gender']=='M']['degree_p']
female = dataset[dataset['gender']=='F']['degree_p']
ttest_ind(male, female)
```

```python
# Since the p value(0.010) is < 0.05 we reject null hypothesis and accept alternative hypothesis
```
