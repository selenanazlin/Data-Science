```python
def QuanQual(dataset):
    Quan = []
    Qual = []
    for columnName in dataset.columns:
        if(dataset[columnName].dtype=='O'):
            Qual.append(columnName)
        else:
            Quan.append(columnName)
    return Quan,Qual
```

```python
# we import "variance_inflation_factor" from statsmodels lib
# this handles the calculation of VIF 
from statsmodels.stats.outliers_influence import variance_inflation_factor
# function is defined with an argument "X"
# X is considered as the dataframe that holds the numarical coloum/feature fro which u need to calculate VIF 
def calc_vif(X):
    # create a empty variable "vif" to hold the o/p
    vif = pd.DataFrame()
    # create a coloum for output variable 
    # "X.columns" pairs each coloumn name directely with its calculated VIF score in the o/p table 
    vif["variables"] = X.columns
    # create a seacond coloum name "VIF"
    # x.shape[1] ----> get the total no of columns in X
    # range(X.shape[1]) ----> loops through column indices
    # X.values ----> converts the dataframe into a raw array
    # variance_inflation_factor(X.values, i) ----> calculate the VIF for the ith column
    vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
    return(vif)
```

```python
calc_vif(dataset[["ssc_p","hsc_p","degree_p","etest_p","mba_p","salary"]])
```
