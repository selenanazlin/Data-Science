```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('50_Startups.csv')
```

```python
dataset
```

```python
dataset.columns
```

```python
dataset = pd.get_dummies (dataset,drop_first=True)
```

```python
indep = dataset[['R&D Spend','Administration','Marketing Spend','State_Florida','State_New York']]
dep = dataset[['Profit']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(indep,dep,test_size = 0.03, random_state = 0)
```

```python
#Standardisation
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.svm import SVR
regressor = SVR(kernel="rbf",C=1000)
regressor.fit(X_train,Y_train)
```

```python
regressor.intercept_
```

```python
regressor.n_support_
```

```python
regressor.support_
```

```python
Y_pred = regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(Y_test,Y_pred)
```

```python
r_score
```

```python
import pickle
filename="finalized_model_svr.sav"
```

```python
pickle.dump(regressor,open(filename,'wb'))
```
